```python

comments_config = "iics.casemix.comments_update_config_test"

df_config = (
    spark.read.table(comments_config)
    .filter(col("is_active") == "1")
)
```

```python
def update_comments(catalog,schema,table,column,comment):
    spark.sql(f"ALTER TABLE {catalog}.{schema}.{table} ALTER COLUMN {column} COMMENT '{comment}'")
```

```python
for i in df_config.collect():
    update_comments(i.catalog_name,i.schema_name,i.table_name,i.column_name,i.column_description)
```
